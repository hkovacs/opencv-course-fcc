```python
import json
import cv2 as cv
import numpy as np
import tensorflow as tf

model = tf.keras.models.load_model("../resources/simpsons-characters-data/model/simpsons_classifier.keras")
with open("../resources/simpsons-characters-data/model/characters.json") as f:
    characters = json.load(f)

IMG_SIZE = (80, 80)

def predict_character(image_path):
    img = cv.imread(image_path)
    if img is None:
        raise FileNotFoundError(f"Could not load image at: {image_path}")
    img = cv.cvtColor(img, cv.COLOR_BGR2RGB)
    img = cv.resize(img, IMG_SIZE)
    tensor = tf.convert_to_tensor(np.reshape(img, (1, *IMG_SIZE, 3)), dtype=tf.float32)
    preds = model(tensor, training=False).numpy()[0]
    idx = np.argmax(preds)
    return f"{characters[idx]} ({preds[idx] * 100:.2f}%)"
```

```python
import matplotlib.pyplot as plt

test_img = "../resources/simpsons-characters-data/kaggle_simpson_testset/lisa/lisa_simpson_8.jpg"

plt.imshow(cv.cvtColor(cv.imread(test_img), cv.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
print(predict_character(test_img))
```
